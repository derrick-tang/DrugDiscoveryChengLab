```python
# PLATE 2-45 to 2-74

# Load data
temp = pd.read_excel('./DrugScreeningData/SC.xlsx', header=None,sheet_name="2-45to2-74")

# Rearrange into rows of 3
drugs = pd.DataFrame(np.reshape(temp.to_numpy(),(-1,3)))

# Add average and std columns
means = drugs.mean(numeric_only=True, axis=1)
stds = drugs.std(numeric_only=True, axis=1)
drugs['average'] = means
drugs['std'] = stds

# Add inhibition column as percentage vs control
drugs['inhib%'] = (drugs.average/drugs.average[31] - 1) * 100
drugs['inhib%std'] = (drugs["std"]/drugs.average[31]) * 100

# Remove controls
drugs = drugs.drop(drugs.index[31])
drugs = drugs.drop(drugs.index[30])

# Name each row
col = list(range(45,74+1))
col = [" 2-" + str(r) for r in col]
drugs['DrugID'] = col
```
